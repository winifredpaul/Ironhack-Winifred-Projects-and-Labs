```python
products =["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
                try:
                    quantity = int(input(f"Enter the quantity of {product} available: "))
                    if quantity < 0:
                        raise ValueError("Invalid quantity! Please enter a non-negative value.")
                    valid_quantity = True
                    inventory[product] = quantity
                except ValueError as error:
                    print(f"Error: {error}")
                    
    return inventory
```

```python
inventory = initialize_inventory(products)
print(inventory)
```

```python
def get_customer_orders(inventory):
    valid_quantity = False
    while not valid_quantity:
        try:
            n = int(input("Enter the number of customer orders: "))
            if n < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
            valid_quantity = True
        except ValueError as error:
                print(f"Error: {error}")

            
    customer_orders = []
    
    for i in range(n):
        valid_quantity = False
        while not valid_quantity:
            customer_order = input(f"Enter the name of product : ").strip().lower()
            if customer_order in inventory and inventory[customer_order] > 0:
                valid_quantity = True
            else:
                print("Enter a valid product name")
        

        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of product to be ordered: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        customer_orders.append(customer_order)
    return customer_orders
```

```python
customer_orders = get_customer_orders(inventory)
print(customer_orders)
```

```python
def total_order_price(customer_orders):
    print("Products in your order:", customer_orders)
    total = 0
    
    for product in customer_orders:
        valid_quantity = False
        while not valid_quantity:
            try:
                price = int(input(f"Enter the price of {product}: "))
                if price < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                total += price
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}") 
    return total
```

```python
total_order_price(customer_orders)
```
