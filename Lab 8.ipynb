```python
clean_customer_insurance_df = combined_df
```

```python
clean_customer_insurance_df = clean_customer_insurance_df.copy()
```

```python
clean_customer_insurance_df.dtypes
```

```python
clean_customer_insurance_df.isnull().sum()
```

```python
# Merge the state columns
clean_customer_insurance_df["state"] = (
    clean_customer_insurance_df["state"]
    .fillna(clean_customer_insurance_df["State"])
)
```

```python
# Merge the gender columns
clean_customer_insurance_df["gender"] = (
    clean_customer_insurance_df["gender"]
    .fillna(clean_customer_insurance_df["Gender"])
)
```

```python
clean_customer_insurance_df.drop(columns=["State", "Gender"], inplace=True)
```

```python
clean_customer_insurance_df.isnull().sum()
```

```python
clean_customer_insurance_df[
    clean_customer_insurance_df["customer_lifetime_value"].isnull()
]
```

```python
clean_customer_insurance_df["customer_lifetime_value"] = (
    clean_customer_insurance_df["customer_lifetime_value"]
    .astype(str)
    .str.replace("%", "", regex=False)
)

clean_customer_insurance_df["customer_lifetime_value"] = pd.to_numeric(
    clean_customer_insurance_df["customer_lifetime_value"],
    errors="coerce"
)
```

```python
clean_customer_insurance_df["customer_lifetime_value"] = (
    clean_customer_insurance_df["customer_lifetime_value"]
    .fillna(clean_customer_insurance_df["customer_lifetime_value"].median())
)
```

```python
clean_customer_insurance_df["customer_lifetime_value"].isnull().sum()
```

```python
numeric_columns = [
    "income",
    "monthly_premium_auto",
    "number_of_open_complaints",
    " total_claim_amount"
]

for col in numeric_columns:
    clean_customer_insurance_df[col] = pd.to_numeric(
        clean_customer_insurance_df[col],
        errors="coerce"
    )
```

```python
for col in numeric_columns:
    clean_customer_insurance_df[col] = clean_customer_insurance_df[col].fillna(
        clean_customer_insurance_df[col].median()
    )
```

```python
clean_customer_insurance_df.isnull().sum()
```

```python
clean_customer_insurance_df["gender"] = clean_customer_insurance_df["gender"].fillna(
    clean_customer_insurance_df["gender"].mode()[0]
)
```

```python
clv_gender_education = customer_marketing_df.pivot_table(
    index="education",
    columns="gender",
    values="customer_lifetime_value",
    aggfunc="mean"
).round(2)

clv_gender_education
```
