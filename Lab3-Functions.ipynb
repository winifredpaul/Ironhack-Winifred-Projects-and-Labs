```python
products =["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
inventory = {}
```

```python
def initialize_inventory(products):
    for product in products:                                            
        product_stock = int(input(f"The quantity of the {product}:") )     
        inventory[product]=  product_stock  
    return inventory
```

```python
initialize_inventory(products)
```

```python

def get_customer_orders():
    customer_orders =set()
    add_more_product = "yes"
    while add_more_product == "yes":
        order= input(f"Enter the name of the product for order ({products}):").strip().lower()
        
        while order not in products:
            print("Select product from the list")
            order= input(f"Enter the name of the product for order ({products}):").strip().lower()
        customer_orders.add(order)   
        add_more_product = input("Add another product? (yes/no): ").strip().lower()
        
        while add_more_product not in ("yes", "no"):
            add_more_product = input("Please type 'yes' or 'no': ").strip().lower()         

    print(customer_orders)
    return customer_orders
    

customer_orders = get_customer_orders()
   


   
```

```python
def update_inventory(customer_orders,inventory):
    for product in customer_orders:
        if inventory[product] >0:
            inventory[product] -= 1
                        
            if inventory[product] == 0:
                print("Product out of stock")

    print(inventory)    

            
update_inventory(customer_orders,inventory)

```

```python
def calculate_order_statistics (customer_orders,products):  
    total_product_ordered = len(customer_orders)                                      #Calculating total order and percentage of product ordered against available order options
    perc_product_order  = ( total_product_ordered/len(products) )*100
 
    return (total_product_ordered, perc_product_order)
        
```

```python
calculate_order_statistics (customer_orders,products)
```

```python
def print_order_statistics(order_statistics):
    total_product_ordered, perc_product_order = order_statistics
    print("Total Products Ordered:", total_product_ordered)
    print("Percentage of Products Ordered:", perc_product_order, "%")
  
```

```python

order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_statistics)
print(order_statistics)
```
