```python
import numpy as np
import pandas as pd
```

```python
customer_insurance_df = pd.read_csv("C:/Users/Winifred/Ironhack/Part time program\Week 4/Lab_7_Customer_insurance_file.csv")
```

```python
customer_insurance_df.info()
```

```python
#Replacing the column label for uniformity in the dataset
customer_insurance_df = customer_insurance_df.rename(columns={"Customer":"customer",
                                                              "ST" : "state", 
                                                              "GENDER" : "gender",
                                                              "Education": "education",
                                                              "Customer Lifetime Value": "customer_lifetime_value",
                                                              "Income" : "income",
                                                              "Monthly Premium Auto" : "monthly_premium_auto",
                                                              "Number of Open Complaints": "number_of_open_complaints",
                                                              "Policy Type" : "policy_type",
                                                              "Vehicle Class": "vehicle_class",
                                                              "Total Claim Amount":" total_claim_amount"                                                        
                                                             })
customer_insurance_df.columns
```

```python
customer_insurance_df.info()
```

```python
customer_insurance_df["gender"].unique()
```

```python
customer_insurance_df["gender"] = customer_insurance_df["gender"].replace ({'Femal':'F',
                                                                            'Male': 'M',
                                                                            'female' : 'F'})
```

```python
customer_insurance_df["gender"].unique()
```

```python
customer_insurance_df["state"].unique()
```

```python
customer_insurance_df["state"] = customer_insurance_df["state"].replace ({'AZ':'Arizona',
                                                                          'WA':'Washington',
                                                                          'Cali':'California'})
                                                                          
```

```python
customer_insurance_df["state"].unique()
```

```python
customer_insurance_df["education"].unique()
```

```python
customer_insurance_df["education"] = customer_insurance_df["education"].replace ({'Bachelors':'Bachelor'})
```

```python
customer_insurance_df["education"].unique()
```

```python
customer_insurance_df["customer_lifetime_value"].unique()
```

```python
customer_insurance_df.loc[:,"customer_lifetime_value"] = (customer_insurance_df["customer_lifetime_value"].str.replace('%', ''))
```

```python
customer_insurance_df["customer_lifetime_value"].unique()
```

```python
customer_insurance_df["vehicle_class"].unique()
```

```python
customer_insurance_df["vehicle_class"] = customer_insurance_df["vehicle_class"].replace ({'Luxury SUV':"Luxury",
                                                                          'Sports Car':"Luxury",
                                                                          'Luxury Car':"Luxury"})
```

```python
customer_insurance_df["vehicle_class"].unique()
```

```python
clean_customer_insurance_df = customer_insurance_df
```

```python
clean_customer_insurance_df = clean_customer_insurance_df.copy()
```

```python
clean_customer_insurance_df.loc[:,"customer_lifetime_value"] = (clean_customer_insurance_df["customer_lifetime_value"].astype(str).str.replace(',', '').str.replace('%', '').astype(float))

clean_customer_insurance_df["customer_lifetime_value"] = pd.to_numeric(
    clean_customer_insurance_df["customer_lifetime_value"],
    errors='coerce')
```

```python
clean_customer_insurance_df.dtypes
```

```python
clean_customer_insurance_df['number_of_open_complaints'] = (clean_customer_insurance_df['number_of_open_complaints'].astype(str) .str.split('/').str[0].str.strip())
```

```python
clean_customer_insurance_df['number_of_open_complaints'] = pd.to_numeric(clean_customer_insurance_df['number_of_open_complaints'],errors='coerce')
```

```python
clean_customer_insurance_df['number_of_open_complaints'] = (clean_customer_insurance_df['number_of_open_complaints'].astype('Int64'))
```

```python
clean_customer_insurance_df.dtypes
```

```python
clean_customer_insurance_df.isnull().sum()
```

```python
#Dropping rows with null values
print("Before dropna:", clean_customer_insurance_df.shape)
```

```python
clean_customer_insurance_df.tail(10)
```

```python
#dataset includes a several blank rows at the bottom of the file.Thus removed rows by comparing it with row that had all cells complete. 
```

```python
formatted_customer_insurance_df = clean_customer_insurance_df.dropna(subset=["customer"])
formatted_customer_insurance_df
```

```python
print("After dropna :", formatted_customer_insurance_df.shape)
```

```python
formatted_customer_insurance_df.isnull().sum()
```

```python
formatted_customer_insurance_df[formatted_customer_insurance_df["gender"].isnull()]
```

```python
formatted_customer_insurance_df = formatted_customer_insurance_df.copy()
```

```python
formatted_customer_insurance_df["gender"] = formatted_customer_insurance_df["gender"].fillna("Unknown")
```

```python
formatted_customer_insurance_df.isnull().sum()
```

```python
median_customer_lifetime_value = formatted_customer_insurance_df["customer_lifetime_value"].median()
print("Median:", median_customer_lifetime_value)
```

```python
Cleaned_customer_insurance_df = formatted_customer_insurance_df.fillna({"customer_lifetime_value":median_customer_lifetime_value})

# Verify: no more nulls in Customer Lifetime Value
print("Nulls in Customer Lifetime Value after fillna:", formatted_customer_insurance_df["customer_lifetime_value"].isnull().sum())
print("Total rows :", Cleaned_customer_insurance_df.shape[0])
```
