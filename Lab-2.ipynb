```python
products =["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
inventory = {}  #Creating a dictionary
```

```python
for product in products:                                             #looking up in the product list
    product_qty= int(input(f"The quantity of the {product}:") )      #creating user input
    inventory[product]= product_qty                                  #adding to dictionary
print(inventory)
    
```

```python
customer_order =set()                                                  #Set creation
```

```python
add_more_product = "yes"

while add_more_product == "yes":
    product_order = input(f"Enter the name of the product: ")

    if product_order:
        customer_order.add(product_order)
    else:
        print(f"Please enter valid product name.")
    
    add_more_product = input(f"Add another product? (yes/no):")
    
print(customer_order)
```

```python
for product in products:                                                        #finding inventory list
    if product in customer_order:
        inventory[product]-=1

print(inventory)
```
