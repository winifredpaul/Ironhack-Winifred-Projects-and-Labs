```python
#Likely categorical columns (based on dtype + meaning): Customer ,ST ,GENDER, Education, Policy Type, Vehicle Class

# These are categorical because they represent labels/groups. We cannot categories customers based on the income or policy value or any column that includes float or integers

#The dataset contains a mix of categorical and numerical variables. Categorical columns such as Gender, Education, Policy Type, and Vehicle Class have a limited number of unique values and are suitable for grouping and segmentation analysis. Numerical variables such as Customer Lifetime Value, Income, and Total Claim Amount show a wide range of values and may contain outliers. The Customer column acts as a unique identifier and is not useful for predictive analysis.
```

```python
numerical_cols = [
    'Customer Lifetime Value',
    'Income',
    'Monthly Premium Auto',
    'Number of Open Complaints',
    'Total Claim Amount'
]

print("Mean:")
print(clean_customer_insurance_df[numerical_cols].mean())
print("\nMedian:")
print(clean_customer_insurance_df[numerical_cols].median())
print("\nMode:")
print(clean_customer_insurance_df[numerical_cols].mode().iloc[0])
print("\nStandard Deviation:")
print(clean_customer_insurance_df[numerical_cols].std())
print("\nQuartiles:")
print(clean_customer_insurance_df[numerical_cols].quantile([0.25, 0.5, 0.75]))
```

```python
personal_auto_df = clean_customer_insurance_df.loc[clean_customer_insurance_df['Policy Type'] == 'Personal Auto']
```

```python
corporate_auto_df = clean_customer_insurance_df.loc[clean_customer_insurance_df['Policy Type'] == 'Corporate Auto']
```

```python
personal_avg_income = personal_auto_df['Income'].mean()
corporate_avg_income = corporate_auto_df['Income'].mean()
```

```python
print(f"Average Income for Personal Auto: {personal_avg_income}")
print(f"Average Income for Corporate Auto: {corporate_avg_income}")
```

```python
if personal_avg_income < corporate_avg_income:
    print("Personal Auto customers have LOWER income than Corporate Auto customers")
elif personal_avg_income > corporate_avg_income:
    print("Personal Auto customers have HIGHER income than Corporate Auto customers")
else:
    print("Both groups have the SAME average income")
```
